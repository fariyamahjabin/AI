```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt
def bfs(tree, root):
    visited = set()
    queue = deque([root])

    while queue:
        vertex = queue.popleft()

        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)

            for child in tree[vertex]:
                if child not in visited:
                    queue.append(child)


# Input Tree
tree = {
    'F': ['A1', 'R'],
    'A1': ['I', 'Y'],
    'R': ['A2'],
    'I': [],
    'Y': [],
    'A2': []
}

print("BFS Traversal:")
bfs(tree, 'F')

# BFS Visualization

G = nx.DiGraph()

for parent, children in tree.items():
    for child in children:
        G.add_edge(parent, child)

pos = {
    'F': (0, 2),
    'A1': (-1, 1),
    'R': (1, 1),
    'I': (-1.5, 0),
    'Y': (-0.5, 0),
    'A2': (1.5, 0)
}

plt.figure(figsize=(8, 5))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    node_color='skyblue',
    font_weight='bold',
    arrows=True
)

plt.title("BFS Tree Visualization")
plt.show()
```

```python
# Implementing DFS (Depth-First Search)
import networkx as nx
import matplotlib.pyplot as plt
def dfs(tree, root, visited=None):
    if visited is None:
        visited = set()

    visited.add(root)
    print(root, end=" ")

    for child in tree[root]:
        if child not in visited:
            dfs(tree, child, visited)


#Input Tree
tree ={
    'F': ['A'],
    'A': ['R'],
    'R': ['I'],
    'I': ['Y'],
    'Y': []
}

print("DFS Traversal Result:")
dfs(tree, 'F')
print()

# DFS Visualization

G = nx.DiGraph()

for parent, children in tree.items():
    for child in children:
        G.add_edge(parent, child)

pos = {
    'F': (0, 2),
    'A': (-1, 1),
    'R': (-1, 0),
    'I': (-1, -1),
    'Y': (-1, -2)
}

plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    node_color='lightgreen',
    font_weight='bold',
    arrows=True
)

plt.title("DFS Tree Visualization")
plt.show()
```

```python
# Implementing DLS (Depth-Limited Search)
import networkx as nx
import matplotlib.pyplot as plt
def dls(tree, root, limit, depth=0):
    print(root, end=" ")

    if depth == limit:
        return

    for child in tree[root]:
        dls(tree, child, limit, depth + 1)


# Input Tree
tree = {
    'F': ['A'],
    'A': ['R'],
    'R': ['I'],
    'I': ['Y'],
    'Y': []
}

print("DLS Traversal Result:")
dls(tree, 'F', 2)
print()

# DLS Visualization

G = nx.DiGraph()

for parent, children in tree.items():
    for child in children:
        G.add_edge(parent, child)

pos = {
    'F': (0, 2),
    'A': (0, 1),
    'R': (0, 0),
    'I': (0, -1),
    'Y': (0, -2)
}

plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    node_color='orange',
    font_weight='bold',
    arrows=True
)

plt.title("DLS Tree Visualization (Depth Limit = 2)")
plt.show()

```

```python
# Implementing IDDFS (Iterative Deepening DFS)
import networkx as nx
import matplotlib.pyplot as plt
def dls(tree, root, limit, depth=0):
    print(root, end=" ")

    if depth == limit:
        return

    for child in tree[root]:
        dls(tree, child, limit, depth + 1)


# Input Tree
tree = {
    'F': ['A'],
    'A': ['R'],
    'R': ['I'],
    'I': ['Y'],
    'Y': []
}

print("IDDFS Traversal Result:")

for limit in range(5):
    print("\nDepth Limit", limit, ": ", end="")
    dls(tree, 'F', limit)

print()


# IDDFS Visualization

G = nx.DiGraph()

for parent, children in tree.items():
    for child in children:
        G.add_edge(parent, child)

pos = {
    'F': (0, 2),
    'A': (0, 1),
    'R': (0, 0),
    'I': (0, -1),
    'Y': (0, -2)
}

plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    node_color='lightblue',
    font_weight='bold',
    arrows=True
)

plt.title("IDDFS Tree Visualization")
plt.show()
```

```python
import heapq
import networkx as nx
import matplotlib.pyplot as plt
def ucs(tree, root):
    visited = set()
    queue = [(0, root)]

    while queue:
        cost, node = heapq.heappop(queue)

        if node not in visited:
            visited.add(node)
            print(node, end=" ")

            for child, edge_cost in tree[node]:
                heapq.heappush(queue, (cost + edge_cost, child))

tree = {
    'F': [('A', 1)],
    'A': [('R', 2)],
    'R': [('I', 3)],
    'I': [('Y', 4)],
    'Y': []
}

print("UCS Traversal Result:")
ucs(tree, 'F')
print()

# UCS Visualization

G = nx.DiGraph()

for parent, children in tree.items():
    for child, cost in children:
        G.add_edge(parent, child, weight=cost)

pos = {
    'F': (0, 2),
    'A': (0, 1),
    'R': (0, 0),
    'I': (0, -1),
    'Y': (0, -2)
}

plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    node_color='lightblue',
    font_weight='bold',
    arrows=True
)

edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels
)

plt.title("UCS Tree Visualization")
plt.show()
```
